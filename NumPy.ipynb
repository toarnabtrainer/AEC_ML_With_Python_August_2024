# **Numpy: Arrays and Matrices**

## NumPy is an extension to the Python Programming language, adding support for large,
## multi-dimensional arrays and with a large-level of mathematical functions to operate
## on those arrays. NumPy is having the full form "Numeric Python".

```python
# importing required module
import numpy as np
```

## **Create arrays**

```python
# creating ndarrays from lists
# note: in one ndarray, all data will have same datatype. (And will be converted if required)
data1 = [100, 300, 200, False, 400, True, 500]
print (data1, len(data1), type(data1))
arr1 = np.array(data1)
print (arr1, len(arr1), type(arr1))
```

```python
data1 = [100, 300, 200.5, 400, True, False, 500]
print (data1, len(data1), type(data1))
arr1 = np.array(data1)
print (arr1, len(arr1), type(arr1))
```

```python
data1 = [100, 300, 200.5, "400", True, False, 500]
print (data1, len(data1), type(data1))
arr1 = np.array(data1)
print (arr1, len(arr1), type(arr1))
```

## **Basic Plots**

```python
# importing all required modules
import numpy as np
import matplotlib.pyplot as plt
# from matplotlib import pyplot as plt
```

```python
# preparing our data for plotting
x = np.linspace(0, 10, 50)
print (x.shape)
print (x, type(x))
```

```python
sinus = np.sin(x)
print (sinus, sinus.shape, type(sinus))
```

```python
plt.plot(x, sinus)
```

```python
plt.plot(x, sinus, "o")
```

```python
plt.plot(x, sinus, "*")
```

```python
plt.plot(x, sinus, "--")
```

```python
plt.plot(x, sinus, "-")
```

```python
help(plt.plot)
```

```python
x = np.linspace(0, 10, 50)
sinus = np.sin(x)
cosinus = np.cos(x)
```

```python
plt.plot(x, cosinus, "o")
```

```python
# rapid multiplot
plt.plot(x, sinus, "-b", x, sinus, "ob", x, cosinus, "-r", x, cosinus, "or")
plt.xlabel("Plotting x-values.")
plt.ylabel("Plotting sin(x) and cos(x).")
plt.title("x vs. sin(x) and cos(x).")
plt.show()
```

```python
# step by step
plt.plot(x, sinus, label = "Sinus", color = "blue", linestyle = "--", linewidth = 2)
plt.plot(x, cosinus, label = "Cosinus", color = "red", linestyle = "-", linewidth = 2)
plt.xlabel("x values plotted...")
plt.ylabel("sin(x) & cos(x) plotted...")
plt.title("x vs. six(x) & cos(x)")
plt.legend(loc = 0)
plt.show()
```
